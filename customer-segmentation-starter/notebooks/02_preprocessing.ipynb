```python
from pathlib import Path
import pandas as pd
from src.data_loader import load_retail_data
from src.preprocess import clean_retail_data
```

```python
DATA_DIR = Path('../data/raw')
OUT_DIR = Path('../data/processed')
OUT_DIR.mkdir(parents=True, exist_ok=True)
candidate_files = list(DATA_DIR.glob('*.xlsx')) + list(DATA_DIR.glob('*.xls')) + list(DATA_DIR.glob('*.csv'))
if not candidate_files:
    raise FileNotFoundError('No raw dataset found in ../data/raw/.')
raw = load_retail_data(candidate_files[0])
raw.head()
```

```python
clean = clean_retail_data(raw)
print('Raw rows:', len(raw))
print('Clean rows:', len(clean))
print('Removed rows:', len(raw) - len(clean))
clean.head()
```

```python
out_path = OUT_DIR / 'online_retail_clean.csv'
clean.to_csv(out_path, index=False)
out_path
```
