# Market Basket Analysis

This notebook creates a basket matrix and extracts association rules.


```python
import os
import sys
from pathlib import Path

def find_project_root(start: Path) -> Path:
    current = start.resolve()
    for candidate in [current, *current.parents]:
        if (candidate / "src").exists():
            return candidate
    raise FileNotFoundError("Could not find project root containing 'src/'")

project_root = find_project_root(Path.cwd())

if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

os.chdir(project_root)

print("Project root:", project_root)
print("src exists:", (project_root / "src").exists())
```

```python
from pathlib import Path
import pandas as pd
from src.market_basket import build_basket_matrix, generate_association_rules
```

```python
from pathlib import Path

clean_path = Path('data/processed/online_retail_clean.csv')
if not clean_path.exists():
    raise FileNotFoundError(f'File not found: {clean_path.resolve()}')

df = pd.read_csv(clean_path, parse_dates=['InvoiceDate'])
df.head()
```

```python
basket = build_basket_matrix(df)
basket.iloc[:5, :10]
```

```python
rules = generate_association_rules(basket, min_support=0.02, metric='lift', min_threshold=1.0)
rules[['antecedents','consequents','support','confidence','lift']].head(20)
```

```python
if not rules.empty:
    top_rules = rules[['antecedents','consequents','support','confidence','lift']].copy()
    top_rules = top_rules.head(10)
    top_rules
else:
    print('No rules found. Lower min_support or filter to a specific country.')
```
